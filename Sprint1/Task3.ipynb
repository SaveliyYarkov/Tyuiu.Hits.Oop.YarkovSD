Описание задачи: Создать базовый класс Student в C#, который будет представлять информацию о студентах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Student: • Атрибуты: Имя (Name), Возраст (Age), Курс (Course). • Методы: o GetInfo(): метод для получения информации о студенте в виде строки. o Study(): метод для вывода сообщения о том, что студент учится. o TakeExam(): метод для вывода сообщения о сдаче экзамена.

Требования к производным классам:

Студент бакалавриата (BachelorStudent): Должен содержать дополнительные атрибуты, такие как Специальность (Specialty). Метод Study() должен быть переопределен для добавления информации о специальности при изучении предмета.
Студент магистратуры (MasterStudent): Должен содержать дополнительные атрибуты, такие как Научный руководитель (ScientificAdvisor). Метод TakeExam() должен быть переопределен для добавления информации о научном руководителе при сдаче экзамена.
Аспирант (PhDStudent) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Тема диссертации (DissertationTopic). Метод Study() должен быть переопределен для добавления информации о теме диссертации при изучении предмета.
Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

```
public class Student
{
    private string name;
    private int age;
    private int course;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public int Course
    {
        get { return course; }
        set
        {
            if (value >= 1)
                course = value;
            else
                throw new ArgumentOutOfRangeException("Курс должен быть не меньше 1!");
        }
    }

    public static int TotalStudents { get; private set; } = 0;

    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        TotalStudents++;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Курс: {Course}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен.");
    }

    public void InteractWith(Student other)
    {
        Console.WriteLine($"{Name} общается с {other.Name}.");
        other.Study();
    }

    public static void DisplayTotalStudents()
    {
        Console.WriteLine($"Общее количество студентов: {TotalStudents}");
    }
}

public class BachelorStudent : Student
{
    private string specialty;

    public string Specialty
    {
        get { return specialty; }
        set { specialty = value; }
    }

    public BachelorStudent(string name, int age, int course, string specialty)
        : base(name, age, course)
    {
        Specialty = specialty;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе бакалавриата по специальности \"{Specialty}\".");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}";
    }
}

public class MasterStudent : Student
{
    private string scientificAdvisor;

    public string ScientificAdvisor
    {
        get { return scientificAdvisor; }
        set { scientificAdvisor = value; }
    }

    public MasterStudent(string name, int age, int course, string scientificAdvisor)
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен под руководством научного руководителя {ScientificAdvisor}.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}";
    }
}

public class PhDStudent : Student
{
    private string dissertationTopic;

    public string DissertationTopic
    {
        get { return dissertationTopic; }
        set { dissertationTopic = value; }
    }

    public PhDStudent(string name, int age, int course, string dissertationTopic)
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} работает над диссертацией на тему \"{DissertationTopic}\" и учится на {Course} курсе аспирантуры.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: {DissertationTopic}";
    }
}

var students = new List<Student>
{
    new BachelorStudent("Макаров Андрей", 20, 2, "Программная инженерия"),
    new MasterStudent("Айтукуджинов Елжас", 24, 1, "Проф. Иванов"),
    new PhDStudent("Кручинин Женя", 28, 2, "Искусственный интеллект")
};

Console.WriteLine("=== ИНФОРМАЦИЯ О СТУДЕНТАХ ===\n");

foreach (var student in students)
{
    Console.WriteLine("--- " + student.GetType().Name + " ---");
    Console.WriteLine(student.GetInfo());
    student.Study();
    student.TakeExam();
    Console.WriteLine();
}

Console.WriteLine("=== ПОЛИМОРФИЗМ В ДЕЙСТВИИ ===\n");

Student[] studentArray = new Student[]
{
    new BachelorStudent("Глеб Уборщиков", 20, 2, "Владение шваброй"),
    new MasterStudent("Олег Симкин", 23, 3, "Доц. Васильева"),
    new PhDStudent("Александр Дунайцев", 26, 3, "Нейронные сети")
};

foreach (var student in studentArray)
{
    Console.WriteLine($"Студент: {student.Name}");
    student.Study();
}

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ СТУДЕНТОВ ===\n");

students[0].InteractWith(students[1]);
students[2].InteractWith(students[0]);

Console.WriteLine();
Student.DisplayTotalStudents();
```
